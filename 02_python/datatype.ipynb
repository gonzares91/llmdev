```python
# 整数の代入
a = 10
b = -5
c = 0

# 整数同士の計算
sum_result = a + b  # 10 + (-5) = 5
product_result = a * b  # 10 * (-5) = -50

print("a:", a)
print("b:", b)
print("c:", c)
print("合計：",sum_result)
print("積：",product_result)
```

```python
#除算の例
division_result = a/3 #結果は結果は 3.3333333333333335（float 型）
integer_division_result =a//3 # 結果は 3（int 型）

print("除算結果：",division_result)
print("整数除算結果：",integer_division_result)
```

```python
#浮動小数点数の代入
x=3.14
y=-0.5
z=2.0 #小数点を付けるとfloat型になる

#浮動小数点数の計算
sum_result=x+y # 3.14 + (-0.5) = 2.64
product_result=x*y # 3.14 * (-0.5) = -1.57

print("x:",x)
print("y:",y)
print("z:",z)
print("合計:", sum_result)
print("積:", product_result)
```

```python
#複素数の代入
c1 = 2 + 3j
c2 = -1 + 0.5j

#複素数の計算
sum_result = c1+c2 #(2+(-1))+(3j+0.5j)=1+3.5j
product_result = c1*c2 #複素数の掛け算

print("c1:",c1)
print("c2:",c2)
print("合計：",sum_result)
print("積：",product_result)
```

```python
#大文字や小文字に変換
text = "Python"
print(text.upper())#'PYTHON'
print(text.lower())#'python'

#部分文字列を置換
text ="Hello, World"
print(text.replace("World","Python")) #'Hello,Python'

#前後の空白や指定した文字を削除
text = "   Hello, World!   "
print(text.strip()) #'Hello, World!'

#部分文字列の最初の位置を検索
text = "Hello, World!"
print(text.find("World")) #7
```

```python
#小数点以下の桁数
pi = 3.14159
print(f"小数点以下の桁数：{pi:2f}") #2桁に指定

#カンマ区切り
large_number = 1234567890
print(f"カンマ区切り: {large_number:,}")

#文字列の出力層
name = "Alice"
print(f"文字列の幅「{name:10}」") #10文字文の幅で表示

#ゼロ埋め
num = 42
print(f"ゼロ埋め：{num:05}")#5桁で表示し足りない部分はゼロで埋める

#パーセント表示
accuracy = 0.856
print(f"％表示：{accuracy:.1%}")

#揃え
name = "Alice"
print(f"左揃え 「{name:<10}」")#左揃えで10文字幅
print(f"右揃え 「{name:>10}」")#右揃えで10文字幅
print(f"中央揃え 「{name:^10}」")#中央揃えで10文字幅

#日時のフォーマット
from datetime import datetime
now = datetime.now()
print(f"現在日時：{now:%Y-%m-%d %H:%M:%S}")
```
