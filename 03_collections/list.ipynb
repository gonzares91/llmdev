```python
#数値を格納したリスト
numbers = [1,2,3,4,5]
print(numbers) #[1,2,3,4,5]

#文字列を格納したリスト
fruits = ["apple","banana","cherry"]
print(fruits) #['apple','banana','cherry']

#異なるデータ型を混在させたリスト
mixed = [1,'apple',3.14,True]
print(mixed)

#空のリスト
empty_list =[]
print(empty_list) #[]
```

```python
#文字列をリストに変換
string_to_list =list("hello")
print(string_to_list) #["h", "e", "l", "l", "o"]

#タプルをリストに変換
tuple_to_list = list((1,2,3))
print(tuple_to_list) #[1,2,3]

#range()関数で生成された範囲をリストに変換
range_to_list = list(range(5))
print(range_to_list) #[0,1,2,3,4]

#list()関数を使用したからのリスト
empty_list = list()
print(empty_list) #[]

```

```python
#Noneで初期化されたリスト
none_list =[None]*5
print(none_list) #[None,None,None,None,None]

#特定の値で初期化されたリスト
zeros =[0]*10
print(zeros)
```

```python
#元のリスト
original = [1,2,3]

#スライスによるコピー
copy1 = original[:]
print(copy1) #[1,2,3]

#list()関数によるコピー
copy2 = list(original)
print(copy2) #[1,2,3]

#copy()メソッドによるコピー
copy3 = original.copy()
print(copy3) #[1,2,3]

```

```python
import copy

#元のリスト（多次元リスト）
original_list = [[1,2],[3,4]]

#浅いコピーを作成
shallow_copied_list = copy.copy(original_list)
#深いコピーを作成
deep_copied_list = copy.deepcopy(original_list)

#浅いコピーのリストを変更
shallow_copied_list[0][0]=100
#深いコピーのリストを変更
deep_copied_list[1][1]=200

#結果の出力
print("元のリスト：",original_list)
print("浅いコピー：",shallow_copied_list)
#浅いコピーの変更が元のリストに影響している

print("深いコピー：",deep_copied_list)
#　→深いコピーの変更は元のリストに影響しない
```

```python
numbers = [0,1,2,3,4,5]

#インデックス1～3の要素を取得（3は含まない）
print(numbers[1:3]) #[1,2]

#最初～インデックス3まで（3は含まない）
print(numbers[:3])#[0,1,2]

#インデックス2～最後まで
print(numbers[2:])#[2,3,4,5]

#全要素を逆順に
print(numbers[::-1]) #[5,4,3,2,1,0]
```

```python
fruits = ["apple","banana"]

#末尾に"cherry"を追加
fruits.append("cherry")
print(fruits)# ["apple", "banana", "cherry"]

# インデックス1の位置に "orange" を挿入
fruits.insert(1,"orange")
print(fruits)# ["apple", "orange", "banana", "cherry"]

# リストに別のリストを結合
fruits.extend(["mango","grape"])
print(fruits) # ["apple", "orange", "banana", "cherry", "mango", "grape"]
```

```python
fruits = ["apple","banana","cherry","banana"]
#指定インデックの要素を削除（インデックス1）
del_item = fruits.pop(1)
print(fruits)#["apple","cherry","banana"]
print(del_item) #banana

#指定の値に一致する最初の要素を削除
fruits.remove("banana")
print(fruits) # ["apple", "cherry"]

#delキーワードで指定の要素を削除（インデックス0）
del fruits[0]
print(fruits) #["charry"]

#全要素を削除してからリストにする
fruits.clear()
print(fruits) #[]
```

```python
list1 = [1,2,3]
list2 = [4,5,6]

#＋演算子をつかって結合
combined = list1 +list2
print(combined)# [1, 2, 3, 4, 5, 6]

#extend()メソッドを使って結合
list1.extend(list2)
print(list1)
```

```python
numbers = [3,1,4,1,5,9,2]

#昇順に並び替え
numbers.sort()
print(numbers) # [1, 1, 2, 3, 4, 5, 9]

#降順に並び替え
numbers.sort(reverse=True)
print(numbers) # [9, 5, 4, 3, 2, 1, 1]

#sorted()関数で並び替えた新しいリストを取得
sorted_numbers = sorted(numbers)
print(sorted_numbers)# [1, 1, 2, 3, 4, 5, 9]
print(numbers)
```

```python
#forループで作成する方法
numbers = []
for i in range(1,6):
    numbers.append(i)
print(numbers) #出力[1,2,3,4,5]
```

```python
#リスト内包表記を使用した方法
numbers = [i for i in range(1,6)]
print(numbers)# 出力: [1, 2, 3, 4, 5]
```

```python
squares = [i**2 for i in range(1,6)]
print(squares)
```

```python
# 偶数はそのまま、奇数は2倍にするリスト
modified_numbers = [i if i % 2 == 0 else i * 2 for i in range(1,6)]
print(modified_numbers) # 出力: [2, 2, 6, 4, 10]
```

```python
even_numbers = [i for i in range(1,11) if i % 2 == 0]
print(even_numbers)
```

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9],
]

#インデックス0行0列の要素にアクセス（1行目1列目）
print(matrix[0][0]) #出力：1

#インデックス2行目1列の要素にアクセス（3行目2列目）
print(matrix[2][1] )#出力：8
```

```python
#3行×3列の2次元リスト
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9],
]

#2次元リストをネストしたループで表示
for row in matrix:
    for element in row:
        print(f"{element:2d}",end="") #改行を出力しない指定
    print() #改行
```

```python
#3×3行列の作成
matrix = [[i*j for j in range(1,4)]for i in range(1,4)]
print(matrix)
```
