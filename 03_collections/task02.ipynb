```python
# 1. 英単語を格納する空リスト
words = []

# 2. アルファベットごとの辞書を作成
alphabet = "abcdefghijklmnopqrstuvwxyz"
count_alphabet ={char:0 for char in alphabet}

# 3. 英単語の入力受付
while True:
    word = input("英単語を入力してください：")

    if word == "":
        break

    words.append(word)

    for char in word:
        if char in count_alphabet:
            count_alphabet[char]+=1

# 4. abc順に並べ替え
words.sort()

# 5. 一覧表示
print(f"入力した英単語：{words}")

for char, count in count_alphabet.items():
    if count > 0:
        print(f"{char}が{count}個ありました")

```
