```python
def is_palindrome(s):
    if s == s[::-1]:
        print(f"「{s}」は回文です！")
    else:
        print(f"「{s}」は回文ではないです。")

is_palindrome("にんてんどううどんてんに")
is_palindrome("Pythonは便利だが遅い")
is_palindrome("no devil lived on")
```

```python
def add(*args):
    result = sum(args)
    print(f"合計:{result}")

add(1,2,3)
add(5,10,15,20)
```

```python
def introduce(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

introduce(name="太郎",age=20,hobby="辞書")
```

```python
def calculate(a,b):
    sum_result = a+b
    product_result= a*b
    return sum_result,product_result
```

```python
result1,result2 = calculate(4,5)
print("足し算計算：",result1)
print("掛け算計算：",result2)
```
